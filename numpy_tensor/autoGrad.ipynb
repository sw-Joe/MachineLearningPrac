Autograd에서 미분(differentiation)

```python
a = torch.tensor([2., 3.], requires_grad=True)
b = torch.tensor([6., 4.], requires_grad=True)
```

```python
Q = 3*a**3 - b**2    # Q=3a^3−b^2
```

```python
# case1. 직접 가중치 전달
external_grad = torch.tensor([1., 1.])
Q.backward(gradient=external_grad)    # 가중치 전달

print(a.grad, b.grad)
```

```python
# case2. 스칼라로 만든 후 미분
loss = Q.sum() # Q1 + Q2 로 합쳐서 스칼라로 만듦
loss.backward()

print(loss)
```

```python
# 수집된 변화도가 올바른지 확인
print(9*a**2 == a.grad)
print(-2*b == b.grad)
```
